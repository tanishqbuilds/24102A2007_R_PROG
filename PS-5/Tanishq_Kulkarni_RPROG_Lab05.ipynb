```r
x <- NA
x
is.na(x)
```

```r
# Checking missing datar

x <- c(7, 11, NA, 15, NA, 20)
x
is.na(x)

which(is.na(x))
sum(is.na(x))
```

```r
# handling missing values

mean(x)

mean(x, na.rm = TRUE)

y <- na.omit(x)
y

mean(y)
```

```r
# NA vs NULL

x <- NA
y <- NULL

x
y

is.na(x)
is.null(x)

is.na(y)
is.null(y)
```

```r
# retrive only compelte observations
x <- c(7, 11, NA, NULL, 15, NA, 20)
x

complete.cases(x)
x[!complete.cases(x)]
```
